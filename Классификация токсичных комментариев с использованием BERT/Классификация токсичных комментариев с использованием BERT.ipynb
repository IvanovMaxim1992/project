# Проект для «Викишоп»


# Классификация токсичных комментариев с использованием BERT

## Описание проекта

Интернет-магазин «Викишоп» добавил функцию пользовательских правок и комментариев к описаниям товаров, что привело к увеличению активности и числу отзывов. Однако для обеспечения дружественной и безопасной среды магазин хочет автоматически выявлять токсичные комментарии и отправлять их на модерацию. Цель — создать инструмент с качеством классификации F1 не ниже 0.75, что позволит обеспечить высокую точность и полноту в выявлении токсичного контента.

## Подготовка

```python
# Библиотеки для работы с данными
import os
import pandas as pd
from collections import Counter

# Библиотеки для обработки текста
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import spacy
from tqdm.notebook import tqdm
from wordcloud import WordCloud

# Библиотеки для визуализации
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Библиотеки для машинного обучения
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression, PassiveAggressiveClassifier
from sklearn.metrics import f1_score, confusion_matrix, classification_report
from sklearn.model_selection import GridSearchCV, train_test_split
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
```

```python
# Загрузка данных
data = pd.read_csv('datasets/toxic_comments.csv', index_col=0)
```

```python
# Проверим общую информацию о данных и наличие пропусков
data.info()
```

```python
# Проверка распределения классов
sns.countplot(x='toxic', data=data)
plt.title('Распределение токсичных и нетоксичных комментариев')
plt.show()
```

```python
# Объединяем все токсичные комментарии в один текст
toxic_text = ' '.join(data[data['toxic'] == 1]['text'])

# Создаём облако слов с отключёнными коллокациями
wordcloud = WordCloud(width=800, height=400, background_color='white', collocations=False).generate(toxic_text)

# Отображаем облако слов
plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Облако слов для токсичных комментариев')
plt.show()
```

```python
# Убираем символы пунктуации и делим текст на слова
words = re.findall(r'\w+', toxic_text.lower())

# Подсчитываем частоту каждого слова
word_counts = Counter(words)

# Получаем 20 самых частотных слов
most_common_words = word_counts.most_common(20)

# Выводим слова и их частоты
print("Топ-20 самых частотных слов в токсичных комментариях:")
for word, count in most_common_words:
    print(f"{word}: {count}")
```

```python
tqdm.pandas()  # Инициализация для использования в pandas
```

```python
# Функция для очистки текста
def clean_text(text):
    text = text.lower()  # Приводим к нижнему регистру
    text = re.sub(r'http\S+', '', text)  # Удаляем ссылки
    text = re.sub(r'[^a-zA-Z]', ' ', text)  # Оставляем только кириллицу
    text = re.sub(r'\s+', ' ', text)  # Удаляем лишние пробелы
    return text.strip()

# Применяем функцию ко всем комментариям
data['cleaned_text'] = data['text'].progress_apply(clean_text)
```

```python
# Загружаем английскую модель SpaCy
nlp = spacy.load('en_core_web_sm')  # Здесь загружаем модель, чтобы она была доступна в функции

# Пример данных
sentence1 = "The striped bats are hanging on their feet for best."
sentence2 = "You should be ashamed of yourself if you behaved like that."
df_my = pd.DataFrame([sentence1, sentence2], columns=['text'])

# Функция для лемматизации с учетом POS-тегов
def lemmatize_text(text):
    doc = nlp(text)
    return ' '.join([token.lemma_ for token in doc if not token.is_stop])

# Тестируем функцию на примерах
df_my['lemmatized_text'] = df_my['text'].progress_apply(lemmatize_text)
display(df_my)
```

```python
# Путь к файлу с лемматизированными данными
lemmatized_data_path = 'datasets/lemmatized_data_english.csv'

if os.path.exists(lemmatized_data_path):
    # Если файл существует, загружаем данные из него
    print("Загружаем лемматизированные данные из файла...")
    data = pd.read_csv(lemmatized_data_path)
else:
    # Если файл не найден, выполняем лемматизацию и сохраняем
    print("Файл не найден, выполняем лемматизацию...")
    data['lemmatized_text'] = data['cleaned_text'].progress_apply(lemmatize_text)
    
    # Сохраняем данные в CSV
    data.to_csv(lemmatized_data_path, index=False)
    print("Лемматизация завершена и данные сохранены в файл.")
```

```python
data.info()
```

```python
# Проверка на наличие NaN в текстовом столбце
print("Проверка на наличие NaN:")
print(data.isnull().sum())  # Выводим количество NaN в каждой колонке

# Проверка на наличие пустых строк
empty_rows = data['lemmatized_text'].str.strip() == ''
print(f"Количество пустых строк в 'lemmatized_text': {empty_rows.sum()}")

# Заполнение NaN и пустых строк
data['lemmatized_text'] = data['lemmatized_text'].fillna('')  # Заполнение NaN
data['lemmatized_text'] = data['lemmatized_text'].replace('', ' ')  # Заменяем пустые строки на пробелы
    
# 1. Разделение данных на обучающую и тестовую выборки с использованием stratify
X = data['lemmatized_text']
y = data['toxic']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=13, stratify=y)


# Проверка распределения классов в выборках
print("Распределение классов в обучающей выборке:")
print(y_train.value_counts(normalize=True))
print("\nРаспределение классов в тестовой выборке:")
print(y_test.value_counts(normalize=True))
```

На основе проведенного анализа данных, включающего визуализацию распределения классов, создание облака слов для токсичных комментариев, предобработку и лемматизацию текста, можно сделать следующие выводы:

1. **Распределение классов**: 
   - Большая часть комментариев (~90%) не токсична, что создаёт дисбаланс классов. При обучении моделей это необходимо учитывать, так как модели могут отдавать предпочтение преобладающему классу (нетоксичным комментариям).

2. **Облако слов для токсичных комментариев**:
   - Часто встречающиеся слова в токсичных комментариях включают сильные ругательства и выражения, что говорит о явном нарушении языка общения и агрессивном поведении. Наибольшие по частоте слова — это ненормативная лексика, а также слова, направленные на унижение или оскорбление других пользователей.

3. **Топ-20 самых частотных слов**:
   - Большинство встречающихся слов в токсичных комментариях включают личные местоимения, наряду с выражениями оскорбительного и негативного характера, что ещё раз подтверждает агрессивный характер этих комментариев.

4. **Предобработка текста**:
   - Были проведены очистка текста от пунктуации и приведение к нижнему регистру. Также проведена лемматизация, чтобы уменьшить количество форм одного и того же слова (например, «hate» и «hates»), что улучшит качество представления текста для модели.

5. **Разделение данных на выборки**:
   - Данные были разделены на обучающую и тестовую выборки с учетом стратификации. Пропорции классов были сохранены в каждой выборке, что позволяет модели лучше адаптироваться к дисбалансу данных.

В дальнейшем, для эффективного выявления токсичных комментариев, необходимо учесть данный дисбаланс классов при обучении моделей. В следующем этапе будет полезно провести подбор гиперпараметров и использовать стратифицированные кросс-валидационные подходы.

## Обучение модели

```python
# Создаем Pipeline для каждой модели
models = {
    'LogisticRegression': Pipeline([
        ('tfidf', TfidfVectorizer(max_features=5000)), 
        ('classifier', LogisticRegression(random_state=42, max_iter=10000))  # Увеличили max_iter
    ]),
    'PassiveAggressiveClassifier': Pipeline([
        ('tfidf', TfidfVectorizer(max_features=5000)),
        ('classifier', PassiveAggressiveClassifier(random_state=42, max_iter=10000))  # Увеличили max_iter
    ])
}

# Параметры для подбора
param_grids = {
    'LogisticRegression': {
        'classifier__C': [0.1, 1.0, 10.0],  # Регуляризация
    },
    'PassiveAggressiveClassifier': {
        'classifier__C': [0.1, 1.0, 10.0],
        'classifier__max_iter': [50, 100]
    }
}

# Словарь для хранения результатов
results = {}

# Обучение и оценка каждой модели
for model_name, model in models.items():
    print(f"Обучение {model_name}...")

    grid_search = GridSearchCV(model, param_grids[model_name], cv=5, scoring='f1', verbose=1)

    try:
        # Обучение модели
        grid_search.fit(X_train, y_train)  # Обучаем на исходных данных

        best_params = grid_search.best_params_
        best_score = grid_search.best_score_

        print(f"Лучшие параметры для {model_name}: {best_params}")
        print(f"Лучший F1-score для {model_name} (на валидации): {best_score:.4f}")

        # Сохраняем результаты
        results[model_name] = {
            'best_params': best_params,
            'best_score': best_score,
            'best_estimator': grid_search.best_estimator_  # Сохраняем лучшую модель
        }

    except Exception as e:
        print(f"Ошибка при обучении {model_name}: {e}")

# Находим лучшую модель по F1-скор на валидации
best_model_name = max(results, key=lambda x: results[x]['best_score'])
best_model = results[best_model_name]['best_estimator']  # Извлекаем уже обученную лучшую модель

# Оцениваем лучшую модель на тестовых данных
y_pred = best_model.predict(X_test)

f1 = f1_score(y_test, y_pred)
print(f"F1-score модели {best_model_name} на тестовых данных: {f1:.4f}\n")
```

```python
# Построение матрицы ошибок
conf_matrix = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', cbar=False,
            xticklabels=np.unique(y_test), yticklabels=np.unique(y_test))
plt.xlabel('Предсказанные метки')
plt.ylabel('Истинные метки')
plt.title(f'Матрица ошибок для {best_model_name}')
plt.show()
```

```python
# Создание отчёта о классификации
target_names = ['Класс 0', 'Класс 1'] 
report = classification_report(y_test, y_pred, target_names=target_names)
print("Отчёт о классификации:\n")
print(report)
```

```python
# Извлекаем модель логистической регрессии
logistic_model = results['LogisticRegression']['best_estimator']

# Попробуем использовать get_feature_names(), если get_feature_names_out() недоступен
try:
    feature_names = best_model.named_steps['tfidf'].get_feature_names_out().tolist()
except AttributeError:
    feature_names = best_model.named_steps['tfidf'].get_feature_names()

# Коэффициенты важности для слов
coefficients = best_model.named_steps['classifier'].coef_[0]

# Создаем DataFrame для визуализации важности признаков
feature_importance = pd.DataFrame({
    'word': feature_names,
    'importance': coefficients
})

# Сортируем признаки по их значению важности и выводим самые важные
top_features = feature_importance.sort_values(by='importance', ascending=False).head(20)

# Построение графика с помощью seaborn
plt.figure(figsize=(10, 6))
sns.barplot(data=top_features, x='importance', y='word')
plt.title('Важность слов для классификации')
plt.xlabel('Коэффициент')
plt.ylabel('Слово')
plt.show()

```
